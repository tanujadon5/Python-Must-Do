```python
# 1.Write a Python program to reverse a string.

def rev_string(s):

  return s[::-1]

s = "hello"

print(rev_string(s))
```

Explanation

1. We have a string, "Hello World", which we want to reverse:

The String to Reverse
txt = "Hello World" [::-1]
print(txt)
Create a slice that starts at the end of the string, and moves backwards.

In this particular example, the slice statement [::-1] means start at the end of the string and end at position 0, move with the step -1, negative one, which means one step backwards.

```python
# 2. Write a Python program to check if a string is a palindrome.

def palindrome(str):
  if str == str[::-1]:
    return True


str = "madam"

print(palindrome(str))
```

```python
# 3. Write a Python program to count the number of vowels in a string.

vowels = "aeiouAEIOU"

def count_vowels(str):
  count = 0

  for char in str:
    if char in vowels:
      count += 1

  return count


#Alternate approach

def count_vowels_2(input_string):
    vowels = "aeiouAEIOU"
    return sum(1 for char in input_string if char in vowels)

    #"For every character that is a vowel, generate 1, then sum all those 1s."

print(count_vowels(str))
print(count_vowels_2(str))

str = "ghysvbooqufpOihgsEkdL"
```

```python
# 4. Write a Python program to find the most frequent character in a string.

def most_frequent_char(str):

  char_count = {}

  for char in str:

    char_count[char] = char_count.get(char, 0) + 1

  return max(char_count,  key=char_count.get)


str = "potato"

print(most_frequent_char(str))
```

max(char_count) does NOT find the most frequent character

This:max(char_count) finds the largest key alphabetically, not the key with the highest count.

For "potato":

p → 2
o → 2
t → 2
a → 1

max(char_count) would return "t" because "t" is alphabetically greatest.

You want the character whose value is maximum.

So write:

return max(char_count, key=char_count.get)
